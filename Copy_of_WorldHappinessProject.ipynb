2. Group the countries according to these regions - Asia, Africa, North America, South America,  Antarctica,  Europe, and Australia.

```python
# Creating dataframes of all countries in the world regionwise
print('\nList of all countries in Asia')
asiadf = pd.read_csv('AsiaCountries.csv')
print(asiadf.to_string())
print('\nList of all countries in Africa')
africadf = pd.read_csv('AfricaCountries.csv')
print(africadf.to_string())
print('\nList of all countries in North America')
namdf = pd.read_csv('NAmericaCountries.csv')
print(namdf.to_string())
print('\nList of all countries in South America')
samdf = pd.read_csv('SAmericaCountries.csv')
print(samdf.to_string())
print('\nList of all countries in Europe')
europedf = pd.read_csv('EuropeCountries.csv')
print(europedf.to_string())
print('\nList of all countries in Australia')
ausdf = pd.read_csv('AustraliaCountries.csv')
print(ausdf.to_string())
print('\n\n')

#Converting to lists
asialist = list(asiadf['Country'])
print('Asia:', end=' ')
print(asialist)
print()
africalist = list(africadf['Country'])
print('Africa:', end=' ')
print(africalist)
print()
namlist = list(namdf['Country'])
print('North America:', end=' ')
print(namlist)
print()
samlist = list(samdf['Country'])
print('South America:', end=' ')
print(samlist)
print()
europelist = list(europedf['Country'])
print('Europe:', end=' ')
print(europelist)
print()
auslist = list(ausdf['Country'])
print('Australia:', end=' ')
print(auslist)
```

```python
#Grouping the countries in the World Happiness report into the correct regions by referring to the lists of countries regionwise

#creating a filter for finding all the Asian countries in the happiness report
fil=hapdf.Country.isin(asialist)
asia_hapdf = hapdf[fil]
print('All the Asia region countries in the World Happiness Report')
print(asia_hapdf.to_string())
print()

#creating a filter for finding all the African countries in the happiness report
fil=hapdf.Country.isin(africalist)
africa_hapdf = hapdf[fil]
print('All the Africa region countries in the World Happiness Report')
print(africa_hapdf.to_string())
print()

#creating a filter for finding all the North American countries in the happiness report
fil=hapdf.Country.isin(namlist)
nam_hapdf = hapdf[fil]
print('All the North America region countries in the World Happiness Report')
print(nam_hapdf.to_string())
print()

#creating a filter for finding all the South American countries in the happiness report
fil=hapdf.Country.isin(samlist)
sam_hapdf = hapdf[fil]
print('All the South America region countries in the World Happiness Report')
print(sam_hapdf.to_string())
print()

#creating a filter for finding all the European countries in the happiness report
fil=hapdf.Country.isin(europelist)
europe_hapdf = hapdf[fil]
print('All the Europe region countries in the World Happiness Report')
print(europe_hapdf.to_string())
print()

#creating a filter for finding all the Australian countries in the happiness report
fil=hapdf.Country.isin(auslist)
aus_hapdf = hapdf[fil]
print('All the Australia region countries in the World Happiness Report')
print(aus_hapdf.to_string())
print()

```

3. Plot the regions (on X-axis) against their average happiness score (the Y-axis)

```python
#Finding the average happiness score per region

regions = ['Asia', 'Africa', 'North America', 'South America', 'Europe', 'Australia']  #list for names of all 6 regions, to plot on X axis
avgscore = []   #list for avg scores of all 6 regions, to plot on Y axis

asia_avg_score = round(asia_hapdf['Score'].mean(),2)
print('\nAverage happiness score of Asia region: ',asia_avg_score)
avgscore.append(asia_avg_score) #adding avg score to list

africa_avg_score = round(africa_hapdf['Score'].mean(),2)
print('\nAverage happiness score of Africa region: ',africa_avg_score)
avgscore.append(africa_avg_score) #adding avg score to list

nam_avg_score = round(nam_hapdf['Score'].mean(),2)
print('\nAverage happiness score of North America region: ',nam_avg_score)
avgscore.append(nam_avg_score) #adding avg score to list

sam_avg_score = round(sam_hapdf['Score'].mean(),2)
print('\nAverage happiness score of South America region: ',sam_avg_score)
avgscore.append(sam_avg_score) #adding avg score to list

europe_avg_score = round(europe_hapdf['Score'].mean(),2)
print('\nAverage happiness score of Europe region: ',europe_avg_score)
avgscore.append(europe_avg_score) #adding avg score to list

aus_avg_score = round(aus_hapdf['Score'].mean(),2)
print('\nAverage happiness score of Australia region: ',aus_avg_score)
avgscore.append(aus_avg_score) #adding avg score to list

print(avgscore)

#Plotting the graph

import matplotlib.pyplot as plt

plt.title('Average Happiness Score per region')
plt.xlabel('Regions', color='red',size='17',weight='bold')
plt.ylabel('Average Score', color='red',size='17',weight='bold')
plt.xticks(rotation=10)
plt.bar(regions,avgscore)
plt.show()
```

4. Find the global average happiness score, and compare each regions average score with it, and print the results as 'Above average', 'Average' , 'Below Average'

```python
world_avg_score = round(hapdf['Score'].mean(),2)
print('\nWorld Average Happiness Score : ',world_avg_score)
for i in range(len(avgscore)):
  if avgscore[i] > world_avg_score:
    print('The happiness score of the ' + regions[i] + ' region(' + str(avgscore[i])+') is \'Above Average\'')
  elif avgscore[i] < world_avg_score:
    print('The happiness score of the ' + regions[i] + ' region(' + str(avgscore[i])+') is \'Below Average\'')
  else:
    print('The happiness score of the ' + regions[i] + ' region(' + str(avgscore[i])+') is \'Average\'')
```
